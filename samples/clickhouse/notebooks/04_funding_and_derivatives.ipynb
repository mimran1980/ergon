## 04 — Funding and derivatives
Funding intervals, mark/index/open interest with provenance; unsupported != zero.

```python
import clickhouse_connect, os
client = clickhouse_connect.get_client(
    host=os.environ.get('CLICKHOUSE_HOST', 'localhost'),
    port=int(os.environ.get('CLICKHOUSE_PORT', '8123')),
    username=os.environ.get('CLICKHOUSE_USER', 'default'),
    password=os.environ.get('CLICKHOUSE_PASSWORD', ''),
    database=os.environ.get('CLICKHOUSE_DATABASE', 'market'),
)
print('clickhouse', client.server_version)
```

```python
funding = client.query("SELECT venue, instrument, rate, event_time_ns FROM funding_rates FINAL ORDER BY event_time_ns LIMIT 20").result_rows
print(funding)
assert funding, 'no funding rows in fixture run'
```

```python
aux = client.query("SELECT table, count() FROM system.tables t LEFT JOIN (SELECT 1) _ ON 1 WHERE database='market' AND name IN ('open_interest','liquidations','mark_prices','index_prices') GROUP BY table").result_rows
print('aux tables present:', aux)
```
